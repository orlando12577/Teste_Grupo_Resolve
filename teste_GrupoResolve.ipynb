```python
qlog["EhPedido"] = (
    qlog["Responsible"]
        .astype(str)
        .str.fullmatch(r"\d+")
)

qlog["ID do Pedido"] = qlog["Responsible"].where(qlog["EhPedido"])

qlog["ID do Pedido"] = qlog["ID do Pedido"].ffill()
```

```python
qlog = qlog[
    ~qlog["EhPedido"]
].copy()
```

```python
qlog = qlog.dropna(how="all")
```

```python
qlog = qlog.drop(
    columns=["Responsible", "EhPedido", "Name"],
    errors="ignore"
)

qlog.head()
```

```python
# Renomear colunas do Q-Log
qlog = qlog.rename(columns={
    "Client": "Cliente",
    "Date": "Data",
    "Part Number": "Número da Peça",
    "Name.1": "Nome",
    "Quantity": "Quantidade",
    "Unit Price": "Preço"
})

# Renomear colunas do System Report
system = system.rename(columns={
    "Order ID": "ID do Pedido",
    "Client": "Cliente",
    "Date": "Data",
    "Part Number": "Número da Peça",
    "Name": "Nome",
    "Quantity": "Quantidade",
    "Price": "Preço"
})
```

```python
qlog["Data"] = pd.to_datetime(
    qlog["Data"],
    dayfirst=True,
    errors="coerce"
)

system["Data"] = pd.to_datetime(
    system["Data"],
    dayfirst=True,
    errors="coerce"
)

qlog["Quantidade"] = pd.to_numeric(
    qlog["Quantidade"],
    errors="coerce"
)

system["Quantidade"] = pd.to_numeric(
    system["Quantidade"],
    errors="coerce"
)

qlog["Preço"] = pd.to_numeric(
    qlog["Preço"],
    errors="coerce"
)

system["Preço"] = pd.to_numeric(
    system["Preço"],
    errors="coerce"
)
```

```python
cancelados = r"cancel|canc|delete|deleted"

qlog = qlog[
    ~qlog["Nome"].str.contains(
        cancelados,
        case=False,
        na=False
    )
]

system = system[
    ~system["Nome"].str.contains(
        cancelados,
        case=False,
        na=False
    )
]
taxas = r"shipping|import|freight|transport|frete"

qlog = qlog[
    ~qlog["Nome"].str.contains(
        taxas,
        case=False,
        na=False
    )
]

system = system[
    ~system["Nome"].str.contains(
        taxas,
        case=False,
        na=False
    )
]

qlog = qlog[qlog["Preço"].notna()]

system = system[system["Preço"].notna()]
```

```python
colunas = [
    "ID do Pedido",
    "Data",
    "Cliente",
    "Número da Peça",
    "Nome",
    "Quantidade",
    "Preço"
]

qlog = qlog[colunas]

system = system[colunas]
```

```python
vendas = pd.concat(
    [system, qlog],
    ignore_index=True
)
vendas.head()

vendas.shape
```

```python
print(vendas.isnull().sum())
```

```python
print("Duplicados:", vendas.duplicated().sum())
```

```python
vendas.info()
```

```python
vendas["Valor Total"] = (
    vendas["Quantidade"] *
    vendas["Preço"]
)
vendas.head()
```

```python
resumo = (
    vendas
    .groupby(
        ["Cliente", "Número da Peça", "Nome"],
        as_index=False
    )
    .agg({
        "Quantidade": "sum",
        "Valor Total": "sum"
    })
)
resumo = resumo.rename(columns={
    "Quantidade": "Quantidade Total",
    "Valor Total": "Valor Total"
})
resumo.head()
```

```python
with pd.ExcelWriter(
    "Resultado_Final.xlsx",
    engine="openpyxl"
) as writer:

    vendas.to_excel(
        writer,
        sheet_name="Vendas Consolidadas",
        index=False
    )

    resumo.to_excel(
        writer,
        sheet_name="Resumo Cliente Produto",
        index=False
    )
```

```python
print("Quantidade de vendas:", len(vendas))
print("Quantidade no resumo:", len(resumo))

print("\nColunas da planilha consolidada:")
print(vendas.columns.tolist())

print("\nColunas do resumo:")
print(resumo.columns.tolist())
```
